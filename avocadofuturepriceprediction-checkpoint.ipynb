```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as mplt 
import random
import seaborn as sns
from prophet import Prophet
```

```python
avocado_dataset = pd.read_csv('avocado.csv')
```

```python
avocado_dataset.head()
```

```python
avocado_dataset.tail(50)
```

```python
avocado_dataset = avocado_dataset.sort_values("Date")
```

```python
avocado_dataset
```

```python
mplt.figure(figsize=(10,10))
mplt.plot(avocado_dataset['Date'], avocado_dataset['AveragePrice'])
```

 # Bar Chart to indicate the number of regions 


```python
mplt.figure(figsize=[25,7])
sns.countplot(x = 'region', data = avocado_dataset)
mplt.xticks(rotation = 80)
```

# Bar Chart to indicate the year


```python
mplt.figure(figsize=[25,12])
sns.countplot(x = 'year', data = avocado_dataset)
mplt.xticks(rotation = 80)
```

```python
avocado_prophet_dataset = avocado_dataset[['Date', 'AveragePrice']] # when changing for prophet make sure its within a double dimensional array
```

```python
avocado_prophet_dataset
```

# date has to have ds as name
# and price has to have y as name for the prophet library to work

```python
avocado_prophet_dataset = avocado_prophet_dataset.rename(columns={'Date':'ds', 'AveragePrice':'y'})
```

```python
avocado_prophet_dataset
```

```python
P = Prophet()
P.fit(avocado_prophet_dataset)
```

```python
# Forcasting into the future
future = P.make_future_dataframe(periods=365)
forecast = P.predict(future)
```

```python
forecast
```

```python
figure = P.plot(forecast, xlabel='Date', ylabel='Price')
```

```python
figure2 = P.plot_components(forecast)
```
